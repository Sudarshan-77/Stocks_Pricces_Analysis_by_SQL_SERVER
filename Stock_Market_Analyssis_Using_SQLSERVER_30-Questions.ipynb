### Q17. Use `CASE WHEN` to categorize each trading record by closing price: Low (below 50), Medium (50 through 150), and High (above 150). Display the symbol, date, closing price, and category.

````python
%%sql 

select 
Symbol, Trade_Date, Close_price,
CASE 
When close_price>50 then 'Low'
When close_price between 50 and 150 then 'Medium'
when close_price>150 then 'High'
else 'Nothing'
END as Category
from Stocks_Prices;
````

### Q19. Calculate the daily percentage change from opening price to closing price. Avoid division-by-zero errors and handle missing opening prices.

````python
%%sql 
with new_cte as (
    select * from Stocks_Prices 
    where open_price is not Null
)
select trade_date, cast(100*(open_price-close_price)/open_price as Decimal(10,2)) as Percentage_Change from Stocks_Prices;
````

## Q21. CTE – Find the Highest Closing Price for Each Stock
Using the original Stocks_Prices dataset, find the highest closing price for each stock symbol during January 2014.

````python
%%sql 

with new_cte as (
select Symbol,close_price,
ROW_NUMBER() OVER (partition by Symbol order by close_price DESC) as RANKING
from Stocks_Prices
where trade_date between '2014-01-02' and '2014-01-30'
)

select * from new_cte where RANKING=1;

````

### Q22. Find records where a stock's closing price is higher than its previous available trading day's closing price. Use a correlated subquery or `EXISTS` approach; compare records within the same stock symbol.

````python
%%sql 

SELECT
    s1.symbol,
    s1.trade_date,
    s1.close_price
FROM Stocks_Prices s1
WHERE s1.close_price >
(
    SELECT TOP 1 s2.close_price
    FROM Stocks_Prices s2
    WHERE s2.symbol = s1.symbol
      AND s2.trade_date < s1.trade_date
    ORDER BY s2.trade_date DESC
);
````

### Q23. Use `ROW_NUMBER()` to number each stock's trading records from highest to lowest closing price. Restart the numbering for each stock symbol.

````python
%%sql 

select Symbol, Close_Price,
ROW_NUMBER() over (partition by Symbol order by close_price ASC) as Stock_Trading_Records
from Stocks_Prices;
````

### Q24. Calculate each stock's average closing price, then use both `RANK()` and `DENSE_RANK()` to rank stock symbols from highest to lowest average closing price.

````python
%%sql 

select Symbol, 
avg(close_price) over (Partition by Symbol order by Close_Price) as Average_Closing_Price,
RANK() over (Partition by Symbol order by Close_Price DESC) as Ranking, 
DENSE_RANK() over (Partition by Symbol order by Close_Price DESC) as Dense_Ranking
from Stocks_Prices;

````

### Q25. Use `NTILE(4)` to divide stock symbols into four groups based on their average closing prices. Show each stock symbol, average closing price, and quartile group.

````python
%%sql

WITH StockAverage AS
(
SELECT
symbol,
AVG(close_price) AS Average_Close_Price
FROM Stocks_Prices
GROUP BY symbol
)
SELECT
symbol,
Average_Close_Price,
NTILE(4) OVER 
(ORDER BY Average_Close_Price DESC) AS Quartile_Group
FROM StockAverage;
````

### Q26. Use `LAG()` to display each stock's previous trading day's closing price alongside its current closing price and date.

````python
%%sql 
select trade_date, Symbol,
Close_Price,
lag(close_Price,1,0) Over (order by trade_date) as Previous_Trading_Date 
from Stocks_Prices;

````

### Q27. Use `LEAD()` to display each stock's next trading day's closing price alongside its current closing price and date.

````python
%%sql 

select 
Symbol, trade_date,Close_Price,
LEAD(Close_Price) over (order by trade_date) as Next_Trading_Date 
from Stocks_Prices;

````

### Q28. Calculate a running total of trading volume for each stock symbol, ordered by trading date.

### Q30. UNPIVOT – Convert Trading Date Columns into Rows
Using the original Stocks_Prices dataset, first create a pivoted table showing each stock symbol's closing prices for January 2, January 3, January 6, January 7, and January 8, 2014.

````python
%%sql
with new_cte as 
(  
Select * from 
(Select trade_date,Symbol, close_price from Stocks_Prices) as Result_set 
pivot (
sum(close_price) for [trade_date]
    in ([2014-01-02],[2014-01-03],[2014-01-06],[2014-01-07])
) as Pivoted_table 
)

select * from new_cte
Unpivot(
    close_price for trade_date
    in([2014-01-02],[2014-01-03],[2014-01-06],[2014-01-07])
) as unpivoted_table;

````

## Notes and reminders

- Use square brackets around SQL Server reserved words or potentially conflicting column names, such as `[date]`, `[open]`, and `[close]`.
- For date ranges, prefer a half-open interval such as `>= '2015-01-01' AND < '2016-01-01'`.
- Stock prices can be `NULL`; decide how to handle missing values when a question requires calculations.
- When comparing daily records, use `PARTITION BY symbol ORDER BY [date]` so one stock is not compared with another.
- If multiple records exist for the same stock on the same date, add a suitable tie-breaker to the window `ORDER BY`.
- `PIVOT` requires an aggregate function. `UNPIVOT` omits rows for `NULL` values in the source columns.
